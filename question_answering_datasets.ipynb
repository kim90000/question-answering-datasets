```python
from transformers import pipeline
import datasets
import torch

# 1. تحميل النموذج (يمكنك اختيار نموذج آخر حسب الحاجة)
# هذا مثال لنموذج سؤال وجواب مدرب مسبقًا
model_name = "deepset/roberta-base-squad2"

# فحص ما إذا كانت وحدة معالجة الرسوميات (GPU) متاحة واستخدامها إذا كان الأمر كذلك
device = 0 if torch.cuda.is_available() else -1

nlp = pipeline('question-answering', model=model_name, tokenizer=model_name, device=device)

# 2. تحميل مجموعة البيانات (مثال: Wikipedia)
dataset_name = "wikipedia"
subset_name = "20220301.en"  # يمكنك اختيار نسخة أحدث إذا كانت متاحة

# تحديد عدد السطور أو الحجم الذي تريد اقتطاعه (مثال: أول 1000 سطر)
num_rows_to_load = 1000

try:
    wiki_dataset = datasets.load_dataset(dataset_name, subset_name, split="train", streaming=True)
    truncated_dataset = wiki_dataset.take(num_rows_to_load)

    # تحويل البيانات المقتطعة إلى قائمة من النصوص
    corpus = [item['text'] for item in truncated_dataset]
    corpus_text = "\n".join(corpus)  # دمج النصوص في سلسلة واحدة كبيرة

    print(f"تم تحميل واقتطاع {num_rows_to_load} سطرًا من مجموعة بيانات Wikipedia.")

except Exception as e:
    print(f"حدث خطأ أثناء تحميل مجموعة البيانات: {e}")
    corpus_text = ""  # في حالة الفشل، اجعل النص فارغًا لتجنب الأخطاء لاحقًا

# 3. وظيفة البحث عن إجابة
def get_answer(question, context):
    """
    تستقبل سؤالًا ونصًا مرجعيًا (context) وتعيد إجابة باستخدام نموذج الذكاء الاصطناعي.
    """
    if not context:
        return "لم يتم تحميل أي بيانات للبحث فيها."

    try:
        result = nlp(question=question, context=context)
        return result['answer']
    except Exception as e:
        return f"حدث خطأ أثناء البحث عن إجابة: {e}"

# 4. مثال للاستخدام
question = "من هو أرشميدس؟"
answer = get_answer(question, corpus_text)

print(f"السؤال: {question}")
print(f"الإجابة: {answer}")

# يمكنك إضافة واجهة مستخدم بسيطة باستخدام مكتبات مثل Gradio أو Streamlit لجعل هذا أكثر تفاعلية.
```

```python
from transformers import pipeline
from datasets import load_dataset

# تحميل نموذج مجاني للإجابة على الأسئلة
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

# تحميل مجموعة بيانات تحتوي على معلومات
# هنا نستخدم "wiki_qa" وهي قاعدة بيانات تحتوي على أسئلة وأجوبة من ويكيبيديا
dataset = load_dataset("wiki_qa")

# استخراج جزء صغير من البيانات لتحديد الحجم
max_size_mb = 1  # الحد الأقصى للحجم بالميجابايت
selected_contexts = []
size_in_bytes = 0

for sample in dataset["train"]:
    context = sample["document_title"] + " " + sample["document_text"]
    size_in_bytes += len(context.encode("utf-8"))
    if size_in_bytes > max_size_mb * 1024 * 1024:
        break
    selected_contexts.append(context)

# دمج النصوص المختارة في مستند واحد
context_data = " ".join(selected_contexts)

# طرح السؤال
question = "Who is Archimedes?"

# الحصول على الإجابة من النموذج
result = qa_pipeline(question=question, context=context_data)

# طباعة الإجابة
print("الإجابة:", result["answer"])
```

```python
# After running pip install haystack-ai "transformers[torch,sentencepiece]"

from haystack import Document
from haystack.components.readers import ExtractiveReader

docs = [
    Document(content="Python is a popular programming language"),
    Document(content="python ist eine beliebte Programmiersprache"),
]

reader = ExtractiveReader(model="deepset/roberta-base-squad2")
reader.warm_up()

question = "What is a popular programming language?"
result = reader.run(query=question, documents=docs)
# {'answers': [ExtractedAnswer(query='What is a popular programming language?', score=0.5740374326705933, data='python', document=Document(id=..., content: '...'), context=None, document_offset=ExtractedAnswer.Span(start=0, end=6),...)]}
```

"question": السؤال المطروح.

"answer": الإجابة الصحيحة.

"document_title": عنوان المقالة التي تتعلق بالسؤال.

```python
from transformers import pipeline
from datasets import load_dataset

# تحميل نموذج مجاني للإجابة على الأسئلة
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

# تحميل مجموعة بيانات WikiQA
dataset = load_dataset("wiki_qa")

# استخراج البيانات المتاحة واستخدام العناوين كمصدر للمعلومات
max_size_mb = 1  # الحد الأقصى للحجم بالميجابايت
selected_contexts = []
size_in_bytes = 0

for sample in dataset["train"]:
    if "document_title" in sample and "answer" in sample:
        context = f"{sample['document_title']}: {sample['answer']}"
        size_in_bytes += len(context.encode("utf-8"))
        if size_in_bytes > max_size_mb * 1024 * 1024:
            break
        selected_contexts.append(context)

# دمج النصوص المختارة في مستند واحد
context_data = " ".join(selected_contexts)

# طرح السؤال
question = "Who is Archimedes?"

# الحصول على الإجابة من النموذج
if context_data.strip():
    result = qa_pipeline(question=question, context=context_data)
    print("الإجابة:", result["answer"])
else:
    print("لم يتم العثور على بيانات كافية للإجابة على السؤال.")
```

```python
import pandas as pd
import datasets

def display_dataset_columns(dataset_name, subset_name=None, num_rows=10):
    """
    Loads a Hugging Face dataset and prints the column names.

    Args:
        dataset_name (str): The name of the dataset on Hugging Face Hub.
        subset_name (str, optional): The name of the subset to load, if applicable. Defaults to None.
        num_rows (int, optional): Number of rows to load from the dataset to extract column names from. Defaults to 10.
    """

    try:
        # Load the dataset using datasets.load_dataset
        if subset_name:
            dataset = datasets.load_dataset(dataset_name, subset_name, split="train", streaming=True)
        else:
            dataset = datasets.load_dataset(dataset_name, split="train", streaming=True)

        # Take the first num_rows to inspect the columns.
        sampled_dataset = dataset.take(num_rows)

        #Convert the sampled dataset to a pandas DataFrame to easily access column names
        df = datasets.Dataset.to_pandas(sampled_dataset)

        # Get the column names
        column_names = df.columns.tolist()


        print(f"Column names for dataset '{dataset_name}' (subset '{subset_name if subset_name else 'None'}'):")
        for col in column_names:
            print(f"- {col}")

    except Exception as e:
        print(f"Error loading or processing the dataset: {e}")


# Example usage:
dataset_name = "rotten_tomatoes"  # Replace with your dataset name
subset_name = None  # Replace with your subset name if needed

display_dataset_columns(dataset_name, subset_name)
```

```python
from transformers import pipeline
import datasets
import torch

# 1. Load the model
model_name = "deepset/roberta-base-squad2"

# Check if GPU is available and use it if so
device = 0 if torch.cuda.is_available() else -1

nlp = pipeline('question-answering', model=model_name, tokenizer=model_name, device=device)

# 2. Load the dataset (example: Wikipedia)
dataset_name = "wikipedia"
subset_name = "20220301.en"  # You can choose a more recent version if available

# Specify the number of rows to load (example: first 1000 rows)
num_rows_to_load = 1000

try:
    wiki_dataset = datasets.load_dataset(dataset_name, subset_name, split="train", streaming=True)
    truncated_dataset = wiki_dataset.take(num_rows_to_load)

    # Convert the truncated data to a list of texts
    corpus = [item['text'] for item in truncated_dataset]
    corpus_text = "\n".join(corpus)  # Combine the texts into a single string

    print(f"Loaded and truncated {num_rows_to_load} rows from the Wikipedia dataset.")

except Exception as e:
    print(f"Error loading the dataset: {e}")
    corpus_text = ""  # In case of failure, make the text empty to avoid errors later

# 3. Function to find an answer
def get_answer(question, context):
    """
    Receives a question and a context text and returns an answer using the AI model.
    """
    if not context:
        return "No data loaded to search."

    try:
        result = nlp(question=question, context=context)
        return result['answer']
    except Exception as e:
        return f"Error finding an answer: {e}"

# 4. Example usage
question = "Who is Archimedes?"
answer = get_answer(question, corpus_text)

print(f"Question: {question}")
print(f"Answer: {answer}")

# You can add a simple user interface using libraries like Gradio or Streamlit to make this more interactive.
```

### الكود المعدل ليعمل مع أي بيانات

```python
from transformers import pipeline
from datasets import load_dataset

# تحميل نموذج مجاني للإجابة على الأسئلة
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

# تحميل أي مجموعة بيانات (هنا نستخدم wiki_qa كمثال، ويمكن تغييره)
dataset = load_dataset("wiki_qa")

# استخراج الحقول النصية تلقائيًا من البيانات المتاحة
def extract_text_fields(sample):
    """Collects all text values ​​from the registry"""
    return " ".join(str(value) for value in sample.values() if isinstance(value, str))

# استخراج البيانات النصية المتاحة
max_size_mb = 1  # الحد الأقصى للحجم بالميجابايت
selected_contexts = []
size_in_bytes = 0

for sample in dataset["train"]:
    context = extract_text_fields(sample)  # استخراج النصوص من أي أعمدة
    size_in_bytes += len(context.encode("utf-8"))
    if size_in_bytes > max_size_mb * 1024 * 1024:
        break
    selected_contexts.append(context)

# دمج النصوص المختارة في مستند واحد
context_data = " ".join(selected_contexts)

# طرح السؤال
question = "glacier caves"

# الحصول على الإجابة من النموذج
if context_data.strip():
    result = qa_pipeline(question=question, context=context_data)
    print("الإجابة:", result["answer"])
else:
    print("لم يتم العثور على بيانات كافية للإجابة على السؤال.")
```

```python
from transformers import pipeline
from datasets import load_dataset

# تحميل نموذج أقوى
qa_pipeline = pipeline("question-answering", model="bert-large-uncased-whole-word-masking-finetuned-squad")

# تحميل 100 سطر بدلًا من 10
dataset = load_dataset("wiki_qa", split="train[:20]")

# استخراج النصوص المتاحة
def extract_text_fields(sample):
    return " ".join(str(value) for value in sample.values() if isinstance(value, str))

# تجميع البيانات النصية
context_data = " ".join(extract_text_fields(sample) for sample in dataset)

# التأكد من أن النص يحتوي على معلومات مفيدة
print("نموذج من البيانات:\n", context_data[:20])

# طرح السؤال
question = "What are glacier caves?"

# الحصول على الإجابة من النموذج
if context_data.strip():
    result = qa_pipeline(question=question, context=context_data)
    print("\n✅ الإجابة:", result["answer"])
else:
    print("❌ لم يتم العثور على بيانات كافية للإجابة على السؤال.")
```

إعادة صياغة السؤال بالكامل لجعل الإجابة أكثر دقة.

```python
from datasets import load_dataset
from collections import Counter
import re
from nltk.corpus import stopwords
import nltk

# تحميل قائمة الكلمات الشائعة (Stopwords) باللغة الإنجليزية
nltk.download("stopwords")
stop_words = set(stopwords.words("english"))

# تحميل داتاسيت (يمكن تغييرها إلى أي داتاسيت أخرى)
dataset = load_dataset("wiki_qa", split="train[:100]")

# استخراج النصوص من الداتاسيت
def extract_text_fields(sample):
    return " ".join(str(value) for value in sample.values() if isinstance(value, str))

# تجميع النصوص في متغير واحد
text_data = " ".join(extract_text_fields(sample) for sample in dataset)

# تنظيف النصوص وإزالة الرموز الخاصة
text_data = re.sub(r"[^\w\s]", "", text_data.lower())

# تقسيم النص إلى كلمات
words = text_data.split()

# إزالة الكلمات الشائعة (Stopwords)
filtered_words = [word for word in words if word not in stop_words]

# حساب تكرار كل كلمة
word_freq = Counter(filtered_words)

# عرض أهم 20 كلمة رئيسية
print("🔹 الكلمات الرئيسية في الداتاسيت:")
for word, freq in word_freq.most_common(20):
    print(f"✅ {word}: {freq} مرات")
```

### شغال الكلمات الرئيسية ف الداتا لاسال عليها

```python
from datasets import load_dataset
from collections import Counter
import re
from nltk.corpus import stopwords
import nltk

# تحميل قائمة الكلمات الشائعة (Stopwords) باللغة الإنجليزية
nltk.download("stopwords")
stop_words = set(stopwords.words("english"))

# تحميل داتاسيت (يمكن تغييرها إلى أي داتاسيت أخرى)
dataset = load_dataset("wiki_qa", split="train[:10]")

# استخراج النصوص من الداتاسيت
def extract_text_fields(sample):
    return " ".join(str(value) for value in sample.values() if isinstance(value, str))

# تجميع النصوص في متغير واحد
text_data = " ".join(extract_text_fields(sample) for sample in dataset)

# تنظيف النصوص وإزالة الرموز الخاصة
text_data = re.sub(r"[^\w\s]", "", text_data.lower())

# تقسيم النص إلى كلمات
words = text_data.split()

# إزالة الكلمات الشائعة (Stopwords)
filtered_words = [word for word in words if word not in stop_words]

# حساب تكرار كل كلمة
word_freq = Counter(filtered_words)

# عرض أهم 20 كلمة رئيسية
print("🔹 الكلمات الرئيسية في الداتاسيت:")
for word, freq in word_freq.most_common(5):
    print(f"✅ {word}: {freq} مرات")
```

```python
from transformers import pipeline
from datasets import load_dataset
import re

# تحميل نموذج أقوى للإجابة على الأسئلة
qa_pipeline = pipeline("question-answering", model="bert-large-uncased-whole-word-masking-finetuned-squad")

# تحميل داتاسيت (الاختيار من 20 سطر كمثال)
dataset = load_dataset("wiki_qa", split="train[:20]")

# استخراج النصوص المتاحة من الأعمدة النصية
def extract_text_fields(sample):
    return " ".join(str(value) for value in sample.values() if isinstance(value, str))

# تجميع النصوص المتاحة في نص واحد
context_data = " ".join(extract_text_fields(sample) for sample in dataset)

# طرح السؤال
question = "motion"

# البحث عن السياق الذي يحتوي على الكلمة المطلوبة
# سنبحث عن كل جملة تحتوي على الكلمة "motion"
context = ""
for sentence in context_data.split("."):
    if "motion" in sentence.lower():  # إذا كانت الكلمة موجودة في الجملة
        context = sentence.strip()  # خذ الجملة كـ سياق

# إذا تم العثور على سياق يحتوي على الكلمة
if context:
    result = qa_pipeline(question=question, context=context)
    print(f"✅ الإجابة: {result['answer']}")
else:
    print("❌ لم يتم العثور على سياق يحتوي على الكلمة المطلوبة.")
```

```python
from transformers import pipeline
from datasets import load_dataset
import re

# تحميل نموذج قوي للإجابة على الأسئلة
qa_pipeline = pipeline("question-answering", model="bert-large-uncased-whole-word-masking-finetuned-squad")

# تحميل داتاسيت (الاختيار من 20 سطر كمثال)
dataset = load_dataset("wiki_qa", split="train[:20]")

# استخراج النصوص المتاحة من الأعمدة النصية
def extract_text_fields(sample):
    return " ".join(str(value) for value in sample.values() if isinstance(value, str))

# تجميع النصوص المتاحة في نص واحد
context_data = " ".join(extract_text_fields(sample) for sample in dataset)

# طرح السؤال
question = "body"

# البحث عن الجمل التي تحتوي على الكلمة المطلوبة (مثل "body")
context = ""
for sentence in context_data.split("."):
    if "body" in sentence.lower():  # إذا كانت الكلمة موجودة في الجملة
        context = sentence.strip()  # خذ الجملة كـ سياق
        break  # أخذ أول جملة تحتوي على الكلمة

# إذا تم العثور على سياق يحتوي على الكلمة
if context:
    # استخدام سياق أطول حول الجملة إذا لزم الأمر
    start_index = context_data.find(context) - 50  # إضافة 50 حرف قبل الجملة للحصول على سياق أكبر
    end_index = context_data.find(context) + len(context) + 50  # إضافة 50 حرف بعد الجملة

    context = context_data[max(0, start_index):min(len(context_data), end_index)]  # تأكيد أنه ضمن حدود النص

    # الحصول على الإجابة من النموذج
    result = qa_pipeline(question=question, context=context)
    print(f"✅ الإجابة: {result['answer']}")
else:
    print("❌ لم يتم العثور على سياق يحتوي على الكلمة المطلوبة.")
```

```python
from transformers import pipeline
from datasets import load_dataset

# تحميل نموذج قوي للإجابة على الأسئلة
qa_pipeline = pipeline("question-answering", model="bert-large-uncased-whole-word-masking-finetuned-squad")

# تحميل داتاسيت (الاختيار من 20 سطر كمثال)
dataset = load_dataset("wiki_qa", split="train[:20]")

# استخراج النصوص المتاحة من الأعمدة النصية
def extract_text_fields(sample):
    return " ".join(str(value) for value in sample.values() if isinstance(value, str))

# تجميع النصوص المتاحة في نص واحد
context_data = " ".join(extract_text_fields(sample) for sample in dataset)

# طرح السؤال
question = "body"  # الكلمة التي ترغب في البحث عنها في النص

# البحث عن الجمل التي تحتوي على الكلمة الخاصة بالسؤال
context = ""
for sentence in context_data.split("."):
    if question.lower() in sentence.lower():  # إذا كانت الكلمة الخاصة بالسؤال موجودة في الجملة
        context = sentence.strip()  # خذ الجملة كـ سياق
        break  # أخذ أول جملة تحتوي على الكلمة

# إذا تم العثور على سياق يحتوي على الكلمة
if context:
    # تحسين السياق المتعلق بالكلمة
    context_data = " ".join([sentence.strip() for sentence in context_data.split(".") if question.lower() in sentence.lower()])

    # الحصول على الإجابة من النموذج
    result = qa_pipeline(question=question, context=context_data)

    # عرض السؤال ونص الإجابة
    print(f"❓ السؤال: {question}")
    print(f"📄 السياق المتعلق بالإجابة: {context_data}")
    print(f"✅ الإجابة: {result['answer']}")
else:
    print("❌ لم يتم العثور على سياق يحتوي على الكلمة المطلوبة.")
```

```python
import json

# دالة لتحميل البيانات من ملف JSON
def load_data(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    return data

# دالة للبحث عن الكلمة في السؤال والإجابة
def search_for_keyword(data, keyword):
    results = []
    for item in data:
        question = item['question']
        answer = item['answer']
        document_title = item['document_title']

        # إذا كانت الكلمة موجودة في السؤال أو الإجابة
        if keyword.lower() in question.lower() or keyword.lower() in answer.lower():
            result = {
                'question': question,
                'answer': answer,
                'document_title': document_title
            }
            results.append(result)
    return results

# تحديد المسار الذي يحتوي على البيانات (تأكد من استبدال المسار الصحيح)
file_path = 'path_to_your_data.json'

# تحميل البيانات من الملف
data = load_data(file_path)

# تحديد الكلمة التي نبحث عنها
keyword = "caves"  # أو أي كلمة أخرى ترغب في البحث عنها

# البحث في البيانات
search_results = search_for_keyword(data, keyword)

# عرض النتائج
for result in search_results:
    print(f"❓ السؤال: {result['question']}")
    print(f"📄 السياق المتعلق بالإجابة: {result['document_title']}")
    print(f"✅ الإجابة: {result['answer']}")
    print("="*50)
```

```python
import os
import json

# دالة لتحميل البيانات من ملف JSON
def load_data(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    return data

# دالة لاقتطاع عدد من الأسطر
def cut_data(data, num_lines):
    # اقتطاع العدد المحدد من الأسطر
    return data[:num_lines]

# دالة لحفظ البيانات في مجلد معين
def save_data_to_folder(data, folder_path, file_name):
    # تأكد من أن المجلد موجود، إذا لم يكن موجودًا، قم بإنشائه
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)

    # تحديد المسار الكامل للملف الذي سيتم حفظ البيانات فيه
    file_path = os.path.join(folder_path, file_name)

    # حفظ البيانات في الملف
    with open(file_path, 'w') as file:
        json.dump(data, file, indent=4)

    print(f"تم حفظ البيانات في {file_path}")

# المسار إلى الملف الذي يحتوي على البيانات
#file_path = 'path_to_your_data.json'  # استبدل بالمسار الفعلي

# المسار إلى المجلد الذي ترغب في حفظ البيانات فيه
folder_path = '/content/a'  # استبدل بالمسار الفعلي للمجلد

# تحميل البيانات من الملف
data = load_data(file_path)

# تحديد عدد الأسطر التي ترغب في اقتطاعها
num_lines = 5  # على سبيل المثال اقتطاع أول 5 أسطر

# اقتطاع البيانات
cut_data_result = cut_data(data, num_lines)

# حفظ البيانات المقتطعة في المجلد المحدد
save_data_to_folder(cut_data_result, folder_path, 'cut_data.json')
```

from collections import Counter
import re
from nltk.corpus import stopwords
import nltk

# تحميل قائمة الكلمات الشائعة (Stopwords) باللغة الإنجليزية
nltk.download("stopwords")
stop_words = set(stopwords.words("english"))

# تحميل داتاسيت (يمكن تغييرها إلى أي داتاسيت أخرى)
dataset = load_dataset("wiki_qa", split="train[:100]")

# استخراج النصوص من الداتاسيت
def extract_text_fields(sample):
    return " ".join(str(value) for value in sample.values() if isinstance(value, str))

# تجميع النصوص في متغير واحد
text_data = " ".join(extract_text_fields(sample) for sample in dataset)

# تنظيف النصوص وإزالة الرموز الخاصة
text_data = re.sub(r"[^\w\s]", "", text_data.lower())

# تقسيم النص إلى كلمات
words = text_data.split()

# إزالة الكلمات الشائعة (Stopwords)
filtered_words = [word for word in words if word not in stop_words]

# حساب تكرار كل كلمة
word_freq = Counter(filtered_words)

# عرض أهم 20 كلمة رئيسية
print("🔹 الكلمات الرئيسية في الداتاسيت:")
for word, freq in word_freq.most_common(20):
    print(f"✅ {word}: {freq} مرات")

import json

# دالة لتحميل البيانات من ملف JSON
def load_data(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    return data

# دالة للبحث عن الكلمة في السؤال والإجابة
def search_for_keyword(data, keyword):
    results = []
    for item in data:
        question = item['question']
        answer = item['answer']
        document_title = item['document_title']

        # إذا كانت الكلمة موجودة في السؤال أو الإجابة
        if keyword.lower() in question.lower() or keyword.lower() in answer.lower():
            result = {
                'question': question,
                'answer': answer,
                'document_title': document_title
            }
            results.append(result)
    return results

# تحديد المسار الذي يحتوي على البيانات (تأكد من استبدال المسار الصحيح)
file_path = 'path_to_your_data.json'

# تحميل البيانات من الملف
data = load_data(file_path)

# تحديد الكلمة التي نبحث عنها
keyword = "caves"  # أو أي كلمة أخرى ترغب في البحث عنها

# البحث في البيانات
search_results = search_for_keyword(data, keyword)

# عرض النتائج
for result in search_results:
    print(f"❓ السؤال: {result['question']}")
    print(f"📄 السياق المتعلق بالإجابة: {result['document_title']}")
    print(f"✅ الإجابة: {result['answer']}")
    print("="*50)

```python
import datasets
import pandas as pd
import os

def save_dataset_subset(dataset_id, output_dir, num_rows=1000, split="train", file_format="csv"):
    """
    Loads a subset of a Hugging Face dataset and saves it to a file in a specified directory.

    Args:
        dataset_id (str): The ID of the dataset on Hugging Face Hub (e.g., "rotten_tomatoes").
        output_dir (str): The directory where the file should be saved.
        num_rows (int, optional): The number of rows to save. Defaults to 1000.
        split (str, optional): The dataset split to use (e.g., "train", "validation", "test"). Defaults to "train".
        file_format (str, optional): The format to save the data in ("csv" or "parquet"). Defaults to "csv".
    """

    try:
        # Load the dataset using datasets.load_dataset with streaming
        dataset = datasets.load_dataset(dataset_id, split=split, streaming=True)

        # Take a subset of the dataset
        subset = dataset.take(num_rows)

        # Convert to Pandas DataFrame
        df = datasets.Dataset.to_pandas(subset)

        # Create the output directory if it doesn't exist
        os.makedirs(output_dir, exist_ok=True)

        # Create the output file path
        file_path = os.path.join(output_dir, f"{dataset_id}_{split}_subset.{file_format}")

        # Save to file based on specified format
        if file_format == "csv":
            df.to_csv(file_path, index=False)  # index=False prevents saving row indices
            print(f"Subset of {dataset_id} saved to {file_path} in CSV format.")
        elif file_format == "parquet":
            df.to_parquet(file_path, index=False)
            print(f"Subset of {dataset_id} saved to {file_path} in Parquet format.")
        else:
            print(f"Error: Invalid file_format '{file_format}'.  Must be 'csv' or 'parquet'.")
            return  # Exit the function if the file format is invalid

    except Exception as e:
        print(f"Error loading or saving the dataset: {e}")


# Example Usage:
dataset_id = "rotten_tomatoes"  # Replace with the dataset you want
output_dir = "data_subsets"  # Replace with the directory where you want to save the data
num_rows = 10  # Replace with the number of rows you want to save
file_format = "csv"  # Or "parquet"

save_dataset_subset(dataset_id, output_dir, num_rows, file_format=file_format)
```

```python
import datasets
import pandas as pd
import os

def save_dataset_subset(dataset_id, output_dir, num_rows=1000, split="train", file_format="csv"):
    """
    Loads a subset of a Hugging Face dataset (streaming) and saves it to a file.

    Args:
        dataset_id (str): Dataset ID on Hugging Face Hub (e.g., "rotten_tomatoes").
        output_dir (str): Directory to save the file.
        num_rows (int, optional): Number of rows to save. Defaults to 1000.
        split (str, optional): Dataset split ("train", "validation", "test"). Defaults to "train".
        file_format (str, optional): "csv" or "parquet". Defaults to "csv".
    """

    try:
        # Load the dataset with streaming
        dataset = datasets.load_dataset(dataset_id, split=split, streaming=True)

        # Create an iterator from the dataset
        dataset_iterator = iter(dataset)

        # Collect data into a list of dictionaries
        data = []
        for _ in range(num_rows):
            try:
                data.append(next(dataset_iterator))
            except StopIteration:
                print("Reached end of dataset before collecting desired number of rows.")
                break  # Stop if we run out of data

        # Convert the list of dictionaries to a Pandas DataFrame
        df = pd.DataFrame(data)

        # Create the output directory if it doesn't exist
        os.makedirs(output_dir, exist_ok=True)

        # Create the output file path
        file_path = os.path.join(output_dir, f"{dataset_id}_{split}_subset.{file_format}")

        # Save to file based on the specified format
        if file_format == "csv":
            df.to_csv(file_path, index=False)
            print(f"Subset of {dataset_id} saved to {file_path} in CSV format.")
        elif file_format == "parquet":
            df.to_parquet(file_path, index=False, engine='pyarrow')
            print(f"Subset of {dataset_id} saved to {file_path} in Parquet format.")
        else:
            print(f"Error: Invalid file_format '{file_format}'. Must be 'csv' or 'parquet'.")

    except Exception as e:
        print(f"Error loading or saving the dataset: {e}")

# Example Usage:
dataset_id = "rotten_tomatoes"
output_dir = "data_subsets"
num_rows = 10
file_format = "csv"  # Or "parquet"

save_dataset_subset(dataset_id, output_dir, num_rows, file_format=file_format)
```

```python
import datasets
from collections import Counter
import re

def extract_keywords(dataset_id, split="train", num_rows=1000, text_column="text", top_n=20):
    """
    Extracts and displays the most frequent keywords from a Hugging Face dataset.

    Args:
        dataset_id (str): The ID of the dataset on Hugging Face Hub.
        split (str, optional): The dataset split to use (e.g., "train"). Defaults to "train".
        num_rows (int, optional): The number of rows to analyze. Defaults to 1000.
        text_column (str, optional): The name of the column containing text. Defaults to "text".
        top_n (int, optional): The number of top keywords to display. Defaults to 20.
    """

    try:
        # Load the dataset (streaming mode)
        dataset = datasets.load_dataset(dataset_id, split=split, streaming=True)

        # Take a subset of the dataset
        subset = dataset.take(num_rows)

        # Combine all text into a single string
        all_text = ""
        for item in subset:
            try:
                all_text += item[text_column] + " " #Concatenate text and add a space
            except KeyError:
                 print(f"Error: The dataset does not have a column named '{text_column}'.")
                 return #Exit function
            except TypeError:
                print(f"Error: Unable to concatenate text in column '{text_column}'.  Check data type.")
                return #Exit Function


        # Clean the text (lowercase, remove punctuation and numbers)
        all_text = all_text.lower()
        all_text = re.sub(r"[^a-zA-Z\s]", "", all_text) #Keep only letters and spaces

        # Tokenize (split into words)
        words = all_text.split()

        # Remove stop words (optional) - You can customize this list
        stop_words = set(["the", "a", "an", "is", "are", "was", "were", "of", "in", "to", "for", "with", "on", "that", "this", "it", "and", "or", "but", "be"])
        words = [word for word in words if word not in stop_words]

        # Count word frequencies
        word_counts = Counter(words)

        # Get the most common words
        most_common_words = word_counts.most_common(top_n)

        # Display the results
        print(f"Top {top_n} keywords in dataset '{dataset_id}' (split '{split}', first {num_rows} rows):")
        for word, count in most_common_words:
            print(f"- {word}: {count}")

    except Exception as e:
        print(f"Error: {e}")



# Example Usage:
dataset_id = "rotten_tomatoes"  # Replace with the ID of the dataset
text_column = "text"  # Replace with the name of the column containing the text
extract_keywords(dataset_id, text_column=text_column)
```

```python
import datasets
import pandas as pd

def display_dataset(dataset_id, split="train", num_rows=10):
    """
    Loads a Hugging Face dataset and displays a sample of the data.

    Args:
        dataset_id (str): The ID of the dataset on Hugging Face Hub (e.g., "rotten_tomatoes").
        split (str, optional): The dataset split to use (e.g., "train", "validation", "test"). Defaults to "train".
        num_rows (int, optional): The number of rows to display. Defaults to 10.
    """

    try:
        # Load the dataset using datasets.load_dataset with streaming
        dataset = datasets.load_dataset(dataset_id, split=split, streaming=True)

        # Take a subset of the dataset
        subset = dataset.take(num_rows)

        # Convert the subset to a Pandas DataFrame
        df = pd.DataFrame(subset)

        # Display the DataFrame
        print(f"Displaying the first {num_rows} rows of dataset '{dataset_id}' (split '{split}'):")
        print(df)  # Use df for display in interactive environment or df.to_string() for printing to console

    except Exception as e:
        print(f"Error loading or displaying the dataset: {e}")


# Example Usage:
dataset_id = "rotten_tomatoes"  # Replace with the ID of the dataset you want to display
display_dataset(dataset_id)
```

```python
import os
import pandas as pd

def load_data(file_path):
    """Load data from a file (can be CSV, JSON, or any other format supported by pandas)."""
    if file_path.endswith('.csv'):
        return pd.read_csv(file_path)
    elif file_path.endswith('.json'):
        return pd.read_json(file_path)
    else:
        raise ValueError("غير مدعوم. فقط ملفات CSV أو JSON مدعومة.")

def save_data_to_folder(data, folder_path, file_name="/content/data_subsets/rotten_tomatoes_train_subset.csv"):
    """Save data partially to a specified folder."""
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)

    # حفظ البيانات إلى ملف في المجلد المحدد
    data.to_csv(os.path.join(folder_path, file_name), index=False)
    print(f"تم حفظ البيانات في المجلد: {folder_path}")

def get_partial_data(data, num_lines):
    "Truncate part of the data."
    return data.head(num_lines)  # استخدام head للحصول على أول N سطرًا من البيانات
```

```python
from sklearn.feature_extraction.text import CountVectorizer

def extract_keywords(data, text_column, top_n=10):
    """Extract keywords from text column"""
    vectorizer = CountVectorizer(stop_words='english', max_features=top_n)
    X = vectorizer.fit_transform(data[text_column])

    # عرض الكلمات الأكثر شيوعًا
    keywords = vectorizer.get_feature_names_out()
    print("الكلمات الرئيسية:")
    print(keywords)
```

```python
def search_for_word(data, search_word, question_column, answer_column):
    """Search for a word in the data and display the associated question and answer.""
    results = data[data[question_column].str.contains(search_word, case=False, na=False)]

    if results.empty:
        print(f"لم يتم العثور على كلمة '{search_word}' في الأسئلة.")
    else:
        print(f"تم العثور على '{search_word}' في الأسئلة. النتائج:")
        for index, row in results.iterrows():
            print(f"\n❓ السؤال: {row[question_column]}")
            print(f"📄 السياق المتعلق بالإجابة: {row[answer_column]}")
            print(f"✅ الإجابة: {row[answer_column]}")
            print("="*50)
```

```python
import os
import pandas as pd

def load_data(file_path):
    """تحميل البيانات من ملف (يمكن أن يكون CSV أو JSON أو أي صيغة أخرى تدعمها pandas)."""
    if not os.path.exists(file_path):
        print(f"الملف {file_path} غير موجود.")
        return None

    try:
        if file_path.endswith('.csv'):
            data = pd.read_csv(file_path)
            print("تم تحميل البيانات بنجاح من CSV.")
        elif file_path.endswith('.json'):
            data = pd.read_json(file_path)
            print("تم تحميل البيانات بنجاح من JSON.")
        else:
            print("تنسيق الملف غير مدعوم. فقط CSV أو JSON مدعومان.")
            return None
    except Exception as e:
        print(f"حدث خطأ أثناء تحميل البيانات: {e}")
        return None
    return data

def save_data_to_folder(data, folder_path, file_name="partial_data.csv"):
    """حفظ البيانات جزئيًا في مجلد محدد."""
    if data is None:
        print("لا توجد بيانات لحفظها.")
        return

    if not os.path.exists(folder_path):
        os.makedirs(folder_path)

    # حفظ البيانات إلى ملف في المجلد المحدد
    data.to_csv(os.path.join(folder_path, file_name), index=False)
    print(f"تم حفظ البيانات في المجلد: {folder_path}")

def get_partial_data(data, num_lines):
    """اقتطاع جزء من البيانات."""
    return data.head(num_lines)  # استخدام head للحصول على أول N سطرًا من البيانات

def display_dataset(data):
    """عرض الداتا سيت بشكل كامل."""
    if data is not None:
        print(data)
    else:
        print("لا توجد بيانات للعرض.")

def extract_keywords(data, text_column, top_n=10):
    """استخراج الكلمات الرئيسية من العمود النصي."""
    if data is None:
        print("لا توجد بيانات لاستخراج الكلمات الرئيسية منها.")
        return

    from sklearn.feature_extraction.text import CountVectorizer

    vectorizer = CountVectorizer(stop_words='english', max_features=top_n)
    X = vectorizer.fit_transform(data[text_column])

    # عرض الكلمات الأكثر شيوعًا
    keywords = vectorizer.get_feature_names_out()
    print("الكلمات الرئيسية:")
    print(keywords)

def search_for_word(data, search_word, question_column, answer_column):
    """البحث عن كلمة في الداتا وعرض السؤال والجواب المرتبط بها."""
    if data is None:
        print("لا توجد بيانات للبحث فيها.")
        return

    results = data[data[question_column].str.contains(search_word, case=False, na=False)]

    if results.empty:
        print(f"لم يتم العثور على كلمة '{search_word}' في الأسئلة.")
    else:
        print(f"تم العثور على '{search_word}' في الأسئلة. النتائج:")
        for index, row in results.iterrows():
            print(f"\n❓ السؤال: {row[question_column]}")
            print(f"📄 السياق المتعلق بالإجابة: {row[answer_column]}")
            print(f"✅ الإجابة: {row[answer_column]}")
            print("="*50)

# تحديد المسار إلى الداتا سيت
file_path = '/content/data_subsets/rotten_tomatoes_train_subset.csv'  # استبدله بالمسار الفعلي
folder_path = '/content/data_subsets'  # استبدله بالمسار الفعلي للمجلد
num_lines_to_load = 10  # عدد الأسطر التي ترغب في تحميلها

# تحميل البيانات
data = load_data(file_path)

# تنزيل جزء من الداتا سيت وحفظه في مجلد
if data is not None:
    partial_data = get_partial_data(data, num_lines_to_load)
    save_data_to_folder(partial_data, folder_path)

# عرض البيانات بالكامل
display_dataset(data)

# استخراج الكلمات الرئيسية من عمود السؤال
extract_keywords(data, 'rock', top_n=5)

# البحث عن كلمة في البيانات (مثال: كلمة "glacier")
search_for_word(data, 'glacier', 'rock', 'answer')
```

```python
import os
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

def load_data(file_path):
    """تحميل البيانات من ملف (يمكن أن يكون CSV أو JSON أو أي صيغة أخرى تدعمها pandas)."""
    if not os.path.exists(file_path):
        print(f"الملف {file_path} غير موجود.")
        return None

    try:
        if file_path.endswith('.csv'):
            data = pd.read_csv(file_path)
            print("تم تحميل البيانات بنجاح من CSV.")
        elif file_path.endswith('.json'):
            data = pd.read_json(file_path)
            print("تم تحميل البيانات بنجاح من JSON.")
        else:
            print("تنسيق الملف غير مدعوم. فقط CSV أو JSON مدعومان.")
            return None
    except Exception as e:
        print(f"حدث خطأ أثناء تحميل البيانات: {e}")
        return None
    return data

def save_data_to_folder(data, folder_path, file_name="partial_data.csv"):
    """حفظ البيانات جزئيًا في مجلد محدد."""
    if data is None:
        print("لا توجد بيانات لحفظها.")
        return

    if not os.path.exists(folder_path):
        os.makedirs(folder_path)

    # حفظ البيانات إلى ملف في المجلد المحدد
    data.to_csv(os.path.join(folder_path, file_name), index=False)
    print(f"تم حفظ البيانات في المجلد: {folder_path}")

def get_partial_data(data, num_lines):
    """اقتطاع جزء من البيانات."""
    return data.head(num_lines)  # استخدام head للحصول على أول N سطرًا من البيانات

def display_dataset(data):
    """عرض الداتا سيت بشكل كامل."""
    if data is not None:
        print(data)
    else:
        print("لا توجد بيانات للعرض.")

def extract_keywords(data, text_column, top_n=10):
    """استخراج الكلمات الرئيسية من العمود النصي."""
    if data is None:
        print("لا توجد بيانات لاستخراج الكلمات الرئيسية منها.")
        return

    # التأكد من أن العمود موجود
    if text_column not in data.columns:
        print(f"العمود {text_column} غير موجود في الداتا سيت.")
        return

    vectorizer = CountVectorizer(stop_words='english', max_features=top_n)
    X = vectorizer.fit_transform(data[text_column])

    # عرض الكلمات الأكثر شيوعًا
    keywords = vectorizer.get_feature_names_out()
    print("الكلمات الرئيسية:")
    print(keywords)

def search_for_word(data, search_word, question_column, answer_column):
    """البحث عن كلمة في الداتا وعرض السؤال والجواب المرتبط بها."""
    if data is None:
        print("لا توجد بيانات للبحث فيها.")
        return

    if question_column not in data.columns or answer_column not in data.columns:
        print(f"العمود {question_column} أو {answer_column} غير موجود في الداتا سيت.")
        return

    results = data[data[question_column].str.contains(search_word, case=False, na=False)]

    if results.empty:
        print(f"لم يتم العثور على كلمة '{search_word}' في الأسئلة.")
    else:
        print(f"تم العثور على '{search_word}' في الأسئلة. النتائج:")
        for index, row in results.iterrows():
            print(f"\n❓ السؤال: {row[question_column]}")
            print(f"📄 السياق المتعلق بالإجابة: {row[answer_column]}")
            print(f"✅ الإجابة: {row[answer_column]}")
            print("="*50)

# تحديد المسار إلى الداتا سيت
file_path = '/content/data_subsets/partial_data.csv'  # استبدله بالمسار الفعلي
folder_path = '/content/data_subsets'  # استبدله بالمسار الفعلي للمجلد
num_lines_to_load = 10  # عدد الأسطر التي ترغب في تحميلها

# تحميل البيانات
data = load_data(file_path)

# تنزيل جزء من الداتا سيت وحفظه في مجلد
if data is not None:
    partial_data = get_partial_data(data, num_lines_to_load)
    save_data_to_folder(partial_data, folder_path)

# عرض البيانات بالكامل
display_dataset(data)

# استخراج الكلمات الرئيسية من عمود النص
extract_keywords(data, 'text', top_n=5)

# البحث عن كلمة في البيانات (مثال: كلمة "rock")
search_for_word(data, 'rock', 'text', 'label')
```

```python
import pandas as pd

def convert_parquet_to_csv(parquet_file, csv_file):
    """تحويل داتا سيت من Parquet إلى CSV."""
    try:
        # قراءة داتا سيت Parquet
        data = pd.read_parquet(parquet_file)

        # حفظ البيانات في ملف CSV
        data.to_csv(csv_file, index=False)
        print(f"تم تحويل البيانات بنجاح إلى: {csv_file}")
    except Exception as e:
        print(f"حدث خطأ أثناء التحويل: {e}")

# مسار ملف Parquet
parquet_file = 'path_to_your_file/wiki_qa.parquet'  # استبدله بالمسار الفعلي لملف Parquet
# مسار ملف CSV الذي سيتم حفظ البيانات فيه
csv_file = 'path_to_save_file/wiki_qa.csv'  # استبدله بالمسار الفعلي لحفظ ملف CSV

# تنفيذ التحويل
convert_parquet_to_csv(parquet_file, csv_file)
```

```python
import pandas as pd

def convert_partial_parquet_to_csv(parquet_file, csv_file, num_rows=10):
    """قراءة 10 سطور من داتا سيت Parquet وتحويلها إلى CSV."""
    try:
        # قراءة داتا سيت Parquet أول 10 سطور فقط
        data = pd.read_parquet(parquet_file, engine='pyarrow')  # تأكد من وجود محرك pyarrow
        partial_data = data.head(num_rows)

        # حفظ البيانات في ملف CSV
        partial_data.to_csv(csv_file, index=False)
        print(f"تم تحويل {num_rows} سطور إلى CSV بنجاح: {csv_file}")
    except Exception as e:
        print(f"حدث خطأ أثناء التحويل: {e}")

# مسار ملف Parquet
parquet_file = '/content/a'  # استبدله بالمسار الفعلي لملف Parquet
# مسار ملف CSV الذي سيتم حفظ البيانات فيه
csv_file = '/content/a'  # استبدله بالمسار الفعلي لحفظ ملف CSV

# تنفيذ التحويل
convert_partial_parquet_to_csv(parquet_file, csv_file, num_rows=10)
```

from transformers import pipeline
from datasets import load_dataset

# تحميل نموذج مجاني للإجابة على الأسئلة
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

# تحميل مجموعة بيانات WikiQA
dataset = load_dataset("wiki_qa")

# استخراج البيانات المتاحة واستخدام العناوين كمصدر للمعلومات
max_size_mb = 1  # الحد الأقصى للحجم بالميجابايت
selected_contexts = []
size_in_bytes = 0

for sample in dataset["train"]:
    if "document_title" in sample and "answer" in sample:
        context = f"{sample['document_title']}: {sample['answer']}"
        size_in_bytes += len(context.encode("utf-8"))
        if size_in_bytes > max_size_mb * 1024 * 1024:
            break
        selected_contexts.append(context)

# دمج النصوص المختارة في مستند واحد
context_data = " ".join(selected_contexts)

# طرح السؤال
question = "من هو أرخميدس؟"

# الحصول على الإجابة من النموذج
if context_data.strip():
    result = qa_pipeline(question=question, context=context_data)
    print("الإجابة:", result["answer"])
else:
    print("لم يتم العثور على بيانات كافية للإجابة على السؤال.")
ما تم تعديله:
استخدام document_title و answer كمصدر للمعلومات بدلًا من document_text غير الموجود.

التحقق من وجود document_title و answer قبل إضافتهما إلى السياق.

from datasets import load_dataset
import re

# تحميل داتاسيت (الاختيار من 20 سطر كمثال)
dataset = load_dataset("wiki_qa", split="train[:20]")

# استخراج النصوص المتاحة من الأعمدة النصية
def extract_text_fields(sample):
    return " ".join(str(value) for value in sample.values() if isinstance(value, str))

# تجميع النصوص المتاحة في نص واحد
context_data = " ".join(extract_text_fields(sample) for sample in dataset)

# طرح السؤال
question = "body"

# البحث عن الجمل التي تحتوي على الكلمة المطلوبة (مثل "body")
context = ""
for sentence in context_data.split("."):
    if "body" in sentence.lower():  # إذا كانت الكلمة موجودة في الجملة
        context = sentence.strip()  # خذ الجملة كـ سياق
        break  # أخذ أول جملة تحتوي على الكلمة

# إذا تم العثور على سياق يحتوي على الكلمة

    start_index = context_data.find(context) - 50  # إضافة 50 حرف قبل الجملة للحصول على سياق أكبر
    end_index = context_data.find(context) + len(context) + 50  # إضافة 50 حرف بعد الجملة

    context = context_data[max(0, start_index):min(len(context_data), end_index)]  # تأكيد أنه ضمن حدود النص

    # الحصول على الإجابة من النموذج

# دالة لتحميل البيانات من ملف JSON
def load_data(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    return data

# دالة للبحث عن الكلمة في السؤال والإجابة
def search_for_keyword(data, keyword):
    results = []
    for item in data:
        question = item['question']
        answer = item['answer']
        document_title = item['document_title']

        # إذا كانت الكلمة موجودة في السؤال أو الإجابة
        if keyword.lower() in question.lower() or keyword.lower() in answer.lower():
            result = {
                'question': question,
                'answer': answer,
                'document_title': document_title
            }
            results.append(result)
    return results

# تحديد المسار الذي يحتوي على البيانات (تأكد من استبدال المسار الصحيح)
file_path = 'path_to_your_data.json'

# تحميل البيانات من الملف
data = load_data(file_path)

# تحديد الكلمة التي نبحث عنها
keyword = "caves"  # أو أي كلمة أخرى ترغب في البحث عنها

# البحث في البيانات

import os
import json

# دالة لتحميل البيانات من ملف JSON
def load_data(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    return data

# دالة لاقتطاع عدد من الأسطر
def cut_data(data, num_lines):
    # اقتطاع العدد المحدد من الأسطر
    return data[:num_lines]

# دالة لحفظ البيانات في مجلد معين
def save_data_to_folder(data, folder_path, file_name):
    # تأكد من أن المجلد موجود، إذا لم يكن موجودًا، قم بإنشائه
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
    
    # تحديد المسار الكامل للملف الذي سيتم حفظ البيانات فيه
    file_path = os.path.join(folder_path, file_name)

    # حفظ البيانات في الملف

    """البحث عن كلمة في الداتا وعرض السؤال والجواب المرتبط بها."""

import os
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

def load_data(file_path):
    """تحميل البيانات من ملف (يمكن أن يكون CSV أو JSON أو أي صيغة أخرى تدعمها pandas)."""
    if not os.path.exists(file_path):
        print(f"الملف {file_path} غير موجود.")
        return None

    try:
        if file_path.endswith('.csv'):
            data = pd.read_csv(file_path)
            print("تم تحميل البيانات بنجاح من CSV.")
        elif file_path.endswith('.json'):
            data = pd.read_json(file_path)
            print("تم تحميل البيانات بنجاح من JSON.")
        else:
            print("تنسيق الملف غير مدعوم. فقط CSV أو JSON مدعومان.")
            return None
    except Exception as e:
        print(f"حدث خطأ أثناء تحميل البيانات: {e}")
        return None
    return data

def save_data_to_folder(data, folder_path, file_name="partial_data.csv"):
    """حفظ البيانات جزئيًا في مجلد محدد."""
    if data is None:
        print("لا توجد بيانات لحفظها.")
        return

    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
    
    # حفظ البيانات إلى ملف في المجلد المحدد
    data.to_csv(os.path.join(folder_path, file_name), index=False)
    print(f"تم حفظ البيانات في المجلد: {folder_path}")

def get_partial_data(data, num_lines):
    """اقتطاع جزء من البيانات."""
    return data.head(num_lines)  # استخدام head للحصول على أول N سطرًا من البيانات

def display_dataset(data):
    """عرض الداتا سيت بشكل كامل."""
    if data is not None:
        print(data)
    else:
        print("لا توجد بيانات للعرض.")

def extract_keywords(data, text_column, top_n=10):
    """استخراج الكلمات الرئيسية من العمود النصي."""
    if data is None:
        print("لا توجد بيانات لاستخراج الكلمات الرئيسية منها.")
        return
    
    # التأكد من أن العمود موجود
    if text_column not in data.columns:
        print(f"العمود {text_column} غير موجود في الداتا سيت.")
        return
    
    vectorizer = CountVectorizer(stop_words='english', max_features=top_n)
    X = vectorizer.fit_transform(data[text_column])
    
    # عرض الكلمات الأكثر شيوعًا
    keywords = vectorizer.get_feature_names_out()
    print("الكلمات الرئيسية:")
    print(keywords)

def search_for_word(data, search_word, question_column, answer_column):
    """البحث عن كلمة في الداتا وعرض السؤال والجواب المرتبط بها."""
    if data is None:
        print("لا توجد بيانات للبحث فيها.")
        return
    
    if question_column not in data.columns or answer_column not in data.columns:
        print(f"العمود {question_column} أو {answer_column} غير موجود في الداتا سيت.")
        return
    
    results = data[data[question_column].str.contains(search_word, case=False, na=False)]
    
    if results.empty:
        print(f"لم يتم العثور على كلمة '{search_word}' في الأسئلة.")
    else:
        print(f"تم العثور على '{search_word}' في الأسئلة. النتائج:")
        for index, row in results.iterrows():
            print(f"\n❓ السؤال: {row[question_column]}")
            print(f"📄 السياق المتعلق بالإجابة: {row[answer_column]}")
            print(f"✅ الإجابة: {row[answer_column]}")
            print("="*50)

# تحديد المسار إلى الداتا سيت
file_path = 'path_to_your_dataset.csv'  # استبدله بالمسار الفعلي
folder_path = '/content/data_subsets'  # استبدله بالمسار الفعلي للمجلد

# تنزيل جزء من الداتا سيت وحفظه في مجلد

    save_data_to_folder(partial_data, folder_path)

display_dataset(data)
